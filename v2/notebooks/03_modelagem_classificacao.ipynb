# 03 — Modelagem: Classificação de Gravidade

Sandbox de experimentação. Para o pipeline final, ver `src/models/train_classification.py`.

```python
import sys, pathlib
sys.path.append(str(pathlib.Path.cwd().parent))
from src.features.build import build_features, load_processed, split_xy
from src.models.train_classification import make_pipeline
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.metrics import classification_report, confusion_matrix
df = build_features(load_processed()).dropna(subset=['gravidade'])
X, y, feats = split_xy(df)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
```

## Baseline — Regressão Logística

```python
pipe = make_pipeline()
pipe.fit(X_train, y_train)
print(classification_report(y_test, pipe.predict(X_test)))
```
